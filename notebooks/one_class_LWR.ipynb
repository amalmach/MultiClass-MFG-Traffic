# 🚗 Single-Class LWR Mean-Field Game

### Interactive Numerical Demonstration

This notebook demonstrates the numerical solution of a single-class
traffic flow model based on Mean-Field Game theory.

The computational framework uses:

- Finite-difference discretization.
- Newton–Krylov nonlinear iterations.
- LGMRES iterative linear solvers.
- Analytical preconditioning.

The experiment reproduces a lightweight configuration of the
original research implementation.

**Scientific references:**

- [Research paper — Transportation Research Part B (2025)](https://doi.org/10.1016/j.trb.2025.103196)
- [PhD thesis — Chapter II (2026)](https://theses.hal.science/tel-05806256/document)

```python
from pathlib import Path
import subprocess
import sys

# ============================================================
# SINGLE-CLASS LWR — CONFIGURABLE PARAMETERS
# ============================================================

# ------------------------------------------------------------
# 1. Physical parameters
# ------------------------------------------------------------

T = 1.0
L = 1.0
u_max = 1.0
rho_jam = 1.0

# ------------------------------------------------------------
# 2. Initial density
# ------------------------------------------------------------

rho_a = 0.2
rho_b = 0.8
gama = 0.15

# ------------------------------------------------------------
# 3. Numerical parameters
# ------------------------------------------------------------

Nx = 15
Nt = 60

CFL = 0.75
EPS = 0.45
mu = 0.0

# ------------------------------------------------------------
# 4. Solver options
# ------------------------------------------------------------

use_preconditioner = True

max_grids = 1

# ============================================================
# Execute the numerical simulation
# ============================================================

ROOT = Path.cwd().resolve().parent

script = (
    ROOT
    / "examples"
    / "one_class_demo"
    / "run_demo.py"
)

command = [
    sys.executable,
    str(script),

    "--T", str(T),
    "--L", str(L),
    "--u-max", str(u_max),
    "--rho-jam", str(rho_jam),

    "--rho-a", str(rho_a),
    "--rho-b", str(rho_b),
    "--gama", str(gama),

    "--nx", str(Nx),
    "--nt", str(Nt),

    "--CFL", str(CFL),
    "--EPS", str(EPS),
    "--mu", str(mu),

    "--preconditioner",
    "on" if use_preconditioner else "off",

    "--max-grids", str(max_grids),
]

subprocess.run(
    command,
    cwd=ROOT,
    check=True,
)

print("LWR simulation completed successfully.")
```

```python
plot_script = ROOT / "examples" / "one_class_demo" / "plot_results.py"

subprocess.run(
    [sys.executable, str(plot_script)],
    cwd=ROOT,
    check=True,
)
```
