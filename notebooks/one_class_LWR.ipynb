# 🚗 Single-Class LWR Mean-Field Game

### Interactive Numerical Demonstration

This notebook demonstrates the numerical solution of a single-class
traffic flow model based on Mean-Field Game theory.

The computational framework uses:

- Finite-difference discretization.
- Newton–Krylov nonlinear iterations.
- LGMRES iterative linear solvers.
- Analytical preconditioning.

The experiment reproduces a lightweight configuration of the
original research implementation.

**Scientific references:**

- [Research paper — Transportation Research Part B (2025)](https://doi.org/10.1016/j.trb.2025.103196)
- [PhD thesis — Chapter II (2026)](https://theses.hal.science/tel-05806256/document)
